### 2. Crea una gráfica de la distribución de la variable 'sepal length (cm)' antes y después del escalado.

```python
from sklearn.preprocessing import MinMaxScaler
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
import pandas as pd

# Cargar datos
iris = load_iris(as_frame=True)
data = iris.data.copy()

# Escalado MinMax
scaler = MinMaxScaler()
data_scaled = pd.DataFrame(scaler.fit_transform(data), columns=iris.feature_names)

# Gráfica comparativa
plt.figure(figsize=(10,5))

# Antes del escalado
plt.hist(data['sepal length (cm)'], bins=20, alpha=0.5, label="Antes (original)")

# Después del escalado
plt.hist(data_scaled['sepal length (cm)'], bins=20, alpha=0.5, label="Después (MinMax)")

plt.title("Distribución de sepal length antes y después del escalado")
plt.xlabel("Valor")
plt.ylabel("Frecuencia")
plt.legend()
plt.grid(True)
plt.show()
```

### 3. Crea una gráfica de dispersión entre 'sepal length (cm)' y 'sepal width (cm)' con los datos escalados por StandardScaler.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_iris
import pandas as pd
import matplotlib.pyplot as plt

# Cargar datos
iris = load_iris(as_frame=True)
data = iris.data.copy()

# Escalado StandardScaler
scaler = StandardScaler()
data_scaled = pd.DataFrame(scaler.fit_transform(data), columns=iris.feature_names)
plt.figure(figsize=(7,5))

plt.scatter(
    data_scaled['sepal length (cm)'],
    data_scaled['sepal width (cm)'],
    alpha=0.7
)

plt.title("Dispersión: Sepal length vs Sepal width (StandardScaler)")
plt.xlabel("Sepal length (escalado)")
plt.ylabel("Sepal width (escalado)")
plt.grid(True)

plt.show()
```
